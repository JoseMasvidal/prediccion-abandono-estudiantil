# Exploración inicial de datos

En este cuaderno realizaremos un **Análisis Exploratorio de Datos (EDA)** del conjunto `student-mat.csv` para entender su estructura y la variable objetivo de rendimiento final (`G3`).

## 1) Importar librerías básicas

Primero cargamos las librerías necesarias:
- **pandas** para manipulación de datos
- **matplotlib** y **seaborn** para visualización

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración estética de gráficos
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
```

## 2) Cargar el dataset

Leemos el archivo `data/student-mat.csv`.

> Nota: este dataset suele estar separado por punto y coma (`;`), por eso usamos `sep=';'`.

```python
ruta_datos = 'data/student-mat.csv'
df = pd.read_csv(ruta_datos, sep=';')

print(f'Dataset cargado correctamente con {df.shape[0]} filas y {df.shape[1]} columnas.')
```

## 3) Primeras filas del dataset

Mostramos las primeras observaciones para entender cómo vienen las columnas y los valores.

## 4) Información general del dataset

- `describe()` para estadísticas descriptivas de variables numéricas

```python
df.info()
```

```python
df.describe()
```

## 5) Distribución de la calificación final (G3)

La variable **G3** representa la calificación final del estudiante.
Visualizamos su distribución para identificar concentración de notas, dispersión y posibles asimetrías.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['G3'], bins=20, kde=True, color='steelblue')
plt.title('Distribución de la calificación final (G3)')
plt.xlabel('G3 (calificación final)')
plt.ylabel('Frecuencia')
plt.show()
```
